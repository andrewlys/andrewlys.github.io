# Polya's Urn

Polya's urn is really cool. For the unitiated:
* At time $0$, put one black ball and one white ball in an urn. 
* At each time $n > 0$, choose one ball from the urn randomly
* Replace the chosen ball, along with a copy of it. 
At time n, we have n + 2 total balls. 

Let $B_n$ be the number of black balls selected, and let 
$$
M_n = \frac{B_{n} + 1}{n + 2}
$$
be the proportion of black balls to total balls in the urn right after time $n$ ($M$ is for martingale). 

We are interested in $\Pr(B_n = k)$ for $0 \le k \le n$. An interesting way to compute this is to first look at the probability of a given path. Let $b_j = 1$ for black and $b_j = 0$ for white. Then let
$$
P(b_1, \ldots, b_n) := \Pr(\text{Ball } 1 = b_1, \ldots, \text{Ball }n = b_n)
$$ 
Suppose we have $b := (b_1, \ldots, b_n)$ such that $|b| = k$. Then we have:
$$
P(b_1, \ldots, b_n) = \prod_{j = 1}^n M_{j - 1}^{b_j} (1 - M_{j-1})^{1 - b_j}
$$
Let's investigate this:
$$
M_{j - 1}^{b_j} (1 - M_{j-1})^{1 - b_j} = \left(\frac{B_{j-1} + 1}{j + 1}\right)^{b_j}\left(\frac{j - B_{j - 1}}{j + 1}\right)^{1 - b_j} = \left(\frac{1 + \sum_{i = 1}^{j - 1} b_i}{j + 1}\right)^{b_j} \left(\frac{1 + \sum_{i = 1}^{j -1} (1 - b_i)}{j + 1}\right)^{1 - b_j}
$$
Of course the denominator is present regardless of $b_j$, so we can pull that out
$$
M_{j - 1}^{b_j} (1 - M_{j-1})^{1 - b_j} = \frac1{j + 1}\left(1 + \sum_{i =1}^{j - 1} b_i\right)^{b_j} \left(1 + \sum_{i = 1}^{j - 1} (1 - b_i) \right)^{ 1 - b_j}
$$

Here's the really clever part. If we look at the sequence $M_{j-1}^{b_j}$, we have something like this:
$$
(\underset{\substack{\uparrow \\ b_1 = 1}}{1}, \underset{\substack{\uparrow \\ b_2 = 0}}{1}, \underset{\substack{\uparrow \\ b_2= 1}}{2}, \ldots, \underset{\substack{\uparrow \\ b_n = 1}}{k})
$$
In other words, at each $b_j = 0$, the entry is $1$, and the non-one entries are incrementing to $k$. We only have $k-1$ non-one entries, and they are precisely $2,3, \ldots, k$. This gives us:
$$
\prod_{j =1}^n M_{j - 1}^{b_j} = 1 \cdot 2 \cdot \ldots \cdot k = k!
$$
We can do a similar thing for $(1 - M_{j - 1})^{1 - b_j}$, and get 
$$
\prod_{j =1}^n (1 - M_{j - 1})^{1 - b_j} = (n - k)!
$$ 
Plugging  in, we get 
$$
P(b_1, \ldots, b_n) = \frac{k! (n - k)!}{(n + 1)!} 
$$
Which means that conditional on $B_n$, the paths are uniform! This is crazy to me! This allows us to very easily compute $\Pr(B_n = k)$. 
$$
\Pr(B_n = k) = \sum_{\substack{b \in \{0,1\}^n \\ |b| = k}} P(b_1, \ldots, b_n) = \sum_{\substack{b \in \{0,1\}^n \\ |b| = k}} \frac{1}{{n \choose k} (n + 1)} = \frac{{n \choose k}}{{n \choose k}(n + 1)} = \frac1{n+1}
$$
Which means that the number of black balls is uniformly distributed. 

This further tells us something rather interesting about $M_n$. We now know that 
$$
\Pr\left(M_n = \frac{k + 1}{n + 2}\right) = \frac1{n + 1}, \qquad 0 \le k \le n
$$
Or that $M_n$ is uniformly distributed on the set 
$$
\left\{\frac1{n+2}, \ldots, \frac{n + 1}{n + 2}\right\}
$$
Which tells us something even more striking! $M_n$ has a limit, and the limit (in distribution!) is $\mathrm{Unif}[0,1]$. 

# A modest extension
Instead of starting the urn with 1 white ball and 1 black ball, let's say that there are $r$ red balls and $b$ blue balls. We define $B_n$ as the number of blue balls chosen, and $M_n$ similarly. 
$$
M_n := \frac{B_n + b}{n + r + b}
$$
Applying the same logic as above, we get that 
$$
M_{j - 1}^{b_j} (1 - M_{j - 1})^{1 - b_j} = \frac{1}{j - 1 + r + b}\left(b + \sum_{i = 1}^{j - 1} b_i\right)^{b_j}\left(r + \sum_{i = 1}^{j - 1} (1 - b_i)\right)^{1 - b_j}
$$
Automatically, for $|b| = k$, we have:
$$
P(b_1, \ldots, b_n) = \frac{b (b + 1) \cdots (b + k - 1) \cdot r (r + 1)\cdots (r + n-k - 1)}{(r+b)(r+b + 1) \cdots (r + b + n - 1)}
$$
Or, writing $x^{\bar{n}} = x(x + 1)\cdots(x + n - 1)$
$$
P(b_1, \ldots, b_n) = \frac{b^{\overline{k}} r^{\overline{n - k}}}{(r + b)^{\bar{n}}}
$$
So, again, conditional on $B_n = k$, all the paths are equally likely. We just as easily get
$$
\Pr(B_n = k) = {n \choose k} \frac{b^{\overline{k}} r^{\overline{n - k}}}{(r + b)^{\bar{n}}}
$$
which, granted isn't as simple as before. However, $M_n$ still has a nice limit. It's intuitively obvious that $M_n$ converges in distribution to something that should have a density, defined by the following limit, where $x:= \frac{k}{n}$:
$$
p(x) = \lim_{n \to \infty} n\Pr\left(M_n = \frac{nx + b}{n + r + b}\right) = \lim_{n \to \infty} n \Pr(B_n = nx)
$$


Since $nx$ shouldn't a priori be an integer, it is a good idea to replace all the factorials with $\Gamma$. 
\begin{align*}
n \Pr(B_n = nx) &= \frac{n\Gamma(n + 1)}{\Gamma(nx + 1)\Gamma(n - nx + 1)} \frac{(\Gamma(b + nx) / \Gamma(b))(\Gamma(r + n - nx) / \Gamma(r))}{\Gamma(r + b + n)/\Gamma(r + b)} 
\end{align*}
Now, we recall: 
$$
B(a, b) = \frac{\Gamma(a)\Gamma(b)}{\Gamma(a + b)}
$$
Which gives us:
\begin{align*}
&= \frac{n^2 \Gamma(n)}{(nx)(n - nx) \Gamma(nx)\Gamma(n - nx)}\frac{(\Gamma(b + nx) / \Gamma(b))(\Gamma(r + n - nx) / \Gamma(r))}{\Gamma(r + b + n)/\Gamma(r + b)} \\
&= x^{-1}(1-x)^{-1} \frac{B(r + b, n)}{B(b, nx)B(r, n(1-x))}\\
\end{align*}
Recall Stirling's formula, which applying to $B(a,b)$, for $a \ll b$
$$
B(a,b) \sim \Gamma(a)b^{-a}
$$
Plugging this in: 
\begin{align*}
&\sim x^{-1}(1-x)^{-1} \frac{\Gamma(r + b) n^{-(r + b)}}{\Gamma(b)(nx)^{-b} \Gamma(r)(n(1-x))^{-r}}\\
&=\frac1{B(r, b)} x^{b - 1}(1-x)^{r - 1} 
\end{align*}

And we finally see that we get 
$$
p(x) = \frac1{B(r,b)} x^{b - 1}(1 - x)^{r - 1}
$$
or that our limit is $\mathrm{Beta}(r,b)$ distributed!

# A modester extension
Now suppose that instead of starting the urn with 1 white ball and 1 black ball, we have $N$ different colors, $c_1, c_2, \ldots, c_N$, and we start the urn with $m_1, m_2, \ldots, m_N$ balls of each color. Suppose we repeat the Polya Urn Process but for this set up. What will the distribution of the limit be? 

After all this set up, there is really only one distribution which generalizes the Beta distribution, as the Beta distribution generalizes the uniform distribution - the Dirichlet distribution!
The way to show this is quite clear, but not yet trivial. 

Letting $C_n^{(j)}$ be how many of color $c_j$ has been selected right after time $n$, and letting $m = m_1+\ldots+ m_N$, then we start to derive, conditional on $C_n^{(j)} = k$: 
$$
P(\text{Ball }1 = c_{j_1}, \ldots, \text{Ball }n = c_{j_n}) = P(c_{j_1}, \ldots, c_{j_n}) = ?
$$ 
It's not exactly clear how we should compute the probability of a given path as we did before. It's underspecified! There's a lot of different ways the non $c_j$ colored balls can be chosen. We should consider the joint distribution of $C^{(j)}$ perhaps. 

We condition on $C_{n}^{(1)} = k_1, \ldots , C_n^{(N - 1)} = k_{N-1}$. Notice that I did not need to include the $N$ th color, since it will automatically be determined: 
$$
C_n^{(N)} = n - \sum_{j = 1}^{N - 1} C_{n}^{(j)} =: k_N
$$
Now, we can get somewhere. 
$$
P(c_{j_1}, \ldots, c_{j_n}) =  \frac{m_1^{\overline{k_1}} \cdot \ldots \cdot m_N^{\overline{k_N}}}{m^{\overline{n}}}
$$
Now, we sum over all the paths that satisfy our condition. Obviously, multinomial coefficients play a role here. 
$$
P(C_{n}^{(1)} = k_1, \ldots, C_{n}^{(N-1)} = k_{N-1}) = {n \choose k_1, \ldots, k_N} \frac{m_1^{\overline{k_1}} \cdot \ldots \cdot m_N^{\overline{k_N}}}{m^{\overline{n}}}
$$

Let's investigate the limiting distribution now. Clearly, it's a multivariate distribution on an $(N-1)$ dimensional space. If we have $x_1, \ldots, x_{N-1}$ where $x_j = \frac{k_j}{n}$, and $x_N = 1 - (x_1 + \ldots + x_{N-1})$ (we need $k_j$ and $n$ to both go to infinity), then similarly to before, we should get:
\begin{align*}
p(x_1, \ldots, x_{N - 1}) &= \lim_{n \to \infty} n^{N - 1} \Pr(C_n^{(1)} = k_1, \ldots, C_{n}^{(N-1)} = k_{N-1})\\
&= \frac{n^N}{k_1 \cdots k_N}\frac{\Gamma(n)}{\Gamma(k_1)\cdots \Gamma(k_N)} \frac{\frac{\Gamma(m_1 + k_1)}{\Gamma(m_1)} \cdots \frac{\Gamma(m_N + k_N)}{\Gamma(m_N)}}{\Gamma(m + n) / \Gamma(m)}\\
&= x_1^{-1} \cdots x_{N}^{-1} \frac{B(m, n)}{B(m_1, k_1)\cdots B(m_N, k_N)}\\
&= x_1^{-1}\cdots x_{N}^{-1} \frac{\Gamma(m)n^{-m}}{\Gamma(m_1)k_1^{-m_1} \cdots \Gamma(m_N)k_N^{-m_N}}\\
&= x_1^{m_1 - 1} \cdots x_N^{m_N - 1} \frac{\Gamma(m)}{\Gamma(m_1) \cdots \Gamma(m_N)}\\
&= \frac{1}{B(m_1, \ldots, m_N)} x_1^{m_1 - 1}\cdots x_N^{m_N - 1}
\end{align*}
where $B(m_1, \ldots, m_N)$ is the multivariate Beta distribution. We confirm our intuition that the limit is Dirichlet distribution. 

# A Bayesian Postscriptum
If you are aware of Bayesian statistics, you may know that the dirichlet distribution provides a family of conjugate priors for the multinomial distribution. I.e. if you have an experiment that can land in one of $N$ different categories with probability $p_j$ for each category, then conditional on $(p_1, \ldots, p_N)$, the result of $n$ experiments is multinomial distribution:
$$
\Pr(X_n = (k_1, \ldots, k_N)| p) = {n \choose k_1, \ldots, k_N} p_1^{k_1} \ldots p_N^{k_N}
$$
The dirichlet distribution is thus conjugate. This is an incredibly important fact in Bayesian statistics, and the dirichlet distribution is used everywhere.
