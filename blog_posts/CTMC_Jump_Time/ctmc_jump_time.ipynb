I was working on a problem from Probability with Martingales, and I ran into some difficulty solving it. Here is the set up: 

Let 
$$
Q = 
\begin{pmatrix}
-a & a \\
b & -b
\end{pmatrix} \qquad a,b > 0
$$
and consider the continuous time markov chain on two states $\{0, 1\}$ defined by the transition probability:
$$
\Pr(X_t | X_0) = P(t) = \exp(t Q)
$$
I had to calculate the following:
$$
\Pr(X_s = 0, 0 \le s \le t)
$$
But I had no idea how to approach this. This isn't a hard markov chain question, and one that if you studied continuous time markov chains, you should very easily how to compute this. However, I was stuck. I played around with this, and noticed that the markov property should force it to satisfy a memorylessness property: 
$$
\Pr(X_s = 0, h \le s \le t + h | X_{h} = 0) = \Pr(X_s = 0 , h \le s \le t + h | X_s = 0, 0 \le s \le h) = \Pr(X_s = 0, 0 \le s \le t | X_0 = 0)
$$
and I remembered that only exponential distribution and geometric distributions are memoryless. I saw the connection, and I reformulated the problem a little. Let 
$$
T := \inf_t \{X_t = 1\}
$$
i.e. $T$ is the first jumping time. Then we can rewrite the above probability as 
$$
\Pr(T > t)
$$
I also didn't know how to calculate it! I was worried about naively integrating over $t$ since stochastic processes are rife with devilish counter examples, so I decided to mess around with it. I considered looking at 
$$
\Pr(T > ds)
$$
for small $ds$. If this set up is nice enough, then we should have
$$
\Pr(T > ds) \approx P_00(ds)
$$
and now we can do some calculations
$$
P(ds) = \exp(ds Q) = \sum_{k = 0}^\infty \frac{(ds Q)^k}{k!} 
$$
Recalling that $ds^2$ is really really small, we should have
$$
P(ds) \approx 1 + ds Q
$$
So 
$$
P_{00}(ds) \approx 1 + ds Q_00 = 1 - a ds
$$
Now, I decided to investigate the same property for exponential distributions. Say we have $T \sim \mathrm{Exp}(\lambda)$, then 
$$
P(T > ds) = e^{-\lambda ds} \approx 1 -\lambda ds
$$
Which established the connection! The $T$ we had above is exponential with parameter $a$. 

The point of this blog post is not this particular fact, which is fairly basic. It's more of a cautionary tail, that heuristic derivations are often very useful, and just monkeying around with differentials can get you very far in mathematics.
